```python
import numpy as np
import ipywidgets as widgets
import IPython.display as ipd
from random import randint

# ======== helpers ========
def gen_tone(amp, freq, tx, phase):
    """Generate tone with raised cosine ramp"""
    tone = amp * np.sin(2*np.pi*freq*tx + phase)

    # ramp (fade in/out)
    Rdur = 50e-3  # ramp duration
    x = np.arange(0, Rdur, tx[1]-tx[0])
    x = np.pi * x / Rdur
    rampUp = (1 + np.cos(x + np.pi)) / 2
    rampDown = np.flip(rampUp)
    wholeramp = np.concatenate((rampUp, np.ones(len(tone)-2*len(x)), rampDown))
    tone = wholeramp * tone
    return tone

def gen_tx(Tdur=1.0, fs=44100):
    """Time axis"""
    dt = 1/fs
    return np.arange(0, Tdur, dt)

# ======== setup ========
fs = 44100
tx = gen_tx(Tdur=1.0, fs=fs)

ref_freq = 2000  # reference tone
ref_amp = 0.1
test_amp = 0.1

# secret starting frequency for test tone (user doesn’t see this)
test_freq_init = ref_freq   # start an octave above

# ======== slider and play functions ========
freq_slider = widgets.IntSlider(
    min=0, max=100, step=1, description='Test Pitch'
)

def play_reference():
    """Play reference tone"""
    tone = gen_tone(ref_amp, ref_freq, tx, 0)
    display(ipd.Audio(tone, rate=fs, autoplay=True))
    # Remove print in real experiment
    print(f"Reference tone: {ref_freq} Hz")  

def play_test(slider_v):
    """Play test tone"""
    f_test = test_freq_init * (2 ** (slider_v/50.0 - 2))
    tone = gen_tone(test_amp, f_test, tx, 0)
    display(ipd.Audio(tone, rate=fs, autoplay=True))
    # Remove print in real experiment
    #print(f"Test tone (hidden freq ~ {f_test:.1f} Hz)")


def show_frequency(b):
    f_test = test_freq_init * (2 ** (s_slide.value/50.0 - 2))
    print(f"Test tone frequency: {f_test:.1f} Hz")


# ======== buttons ========
widgets.interact(play_reference())


#c_slide = widgets.IntSlider(min=0,max=180,step=10,description='phase diff')
s_slide = widgets.IntSlider(min=randint(5,15),max=randint(110,120),step=1,description='test tone freq',layout=widgets.Layout(width='800px'))  # set desired width

widgets.interact(play_test, slider_v=s_slide)

button_show = widgets.Button(description="Show Test Frequency")
button_show.on_click(show_frequency)


# ======== layout ========
widgets.VBox([button_show])
```
