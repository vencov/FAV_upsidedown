```python
# example on SDT, calcualtes dprime and C from probabilities of hit and false alarm (FA)

import numpy as np

import scipy.stats as st


# set the probability of HIT:
Phit = 0.9 
# set the probability of false alarm (FA)
PFA = 0.2


'''
For the calculation of dprime, we assume that noise and signal probabilities have the same distributions and the same 
and that they have the same dtandard deviation
Therefore, dprime is the distance between the mean values, which is equal to the distances between the z-values of hits and false alarms: zhit-zFA
'''


zhit = st.norm.ppf(Phit)
zFA = st.norm.ppf(PFA)

dprime = zhit - zFA  # dprime from z-values

'''
For the calculation of C value describing the strategy of the listener, we take the equation: C = B - 1/2*dprime
Because z-value for FA is equal to -B and dprime is given to zhit-zFA, we get: C = -0.5*(zhit + zFA)
'''
C = -0.5*(zhit + zFA)

print(f'dprime = {dprime:.2f}, C = {C:.2f}')




```
